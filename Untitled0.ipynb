```python
def sub_sets(string,index,result):
    if index==len(string):
        print(result)
        return
    result.append(string[index])
    sub_sets(string,index+1,result)
    result.pop()
    sub_sets(string,index+1,result)
string=[30,20,10]
index=0
result=[]
sub_sets(string,index,result)
```
