```python
import re
text="this is is a simple simple text of of regrex"
#\1 refers back to whatever group 1 matched-- a repeated word check
pattern=r"\b(\w+)\s+\1\b"
repeats =re.findall(pattern,text,flags=re.IGNORECASE)
print("Repeated words found:",repeats)
```
