```python
def sub_sets(string,index,result):
    if index==len(string):
        print(result)
        return
    result.append(string[index])
    sub_sets(string,index+1,result)
    result.pop()
    sub_sets(string,index+1,result)
string=[30,20,10]
index=0
result=[]
sub_sets(string,index,result)
```

```python
import re
email=input("Enter Email:")
pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'
if re.fullmatch(pattern, email):
  print("Valid email")
else:
  print("Invalid email")
```
